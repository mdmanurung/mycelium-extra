# QC exploration (SIMULATED)

Library sizes and RIN by run. Exploration only; results come from `analysis/`.

```python
import csv
with open('../data/processed/vaccine-cohort/sample_metadata.tsv') as handle:
    rows = list(csv.DictReader(handle, delimiter='\t'))
for run in sorted(set(r['run_id'] for r in rows)):
    rins = [float(r['rin']) for r in rows if r['run_id'] == run]
    print(run, len(rins), min(rins), max(rins))
```
